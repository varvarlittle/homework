# Control Flow в Python



Python підтримує всі найпоширеніші інструкції керування потоком (Control Flow) типу "якщо ..., то ...", а також ітерування по набору значень (цикли).

**Важливо**: при використанні інструкцій керування потоком у нас будуть зʼявлятись вкладені блоки.   
Аби виділити вкладені блоки в Python ми використовуємо **"відступи"**. Відступи - це елемент синтаксису.  
Якщо ви пропустили десь виступ - це буде помилкою, про що вам повідомить компілятор.

## Умовні конструкції. If...else statements

Оператор `if` обчислює умову (вираз, результатом якого є `True` або `False`). Якщо умова істинна, блок коду всередині `if` виконується. Якщо умова хибна, блок коду пропускається.

````python
# Програма, що ілюструє роботу оператора if
i = 25

if i > 15:
    print("значення більше 15")
    print(i+100)
    print("Блок не всередині if")

print("Блок не всередині if")
print(10-1)
i = 10
print(i)
````

## Цикли (loops)

Уявіть, що вам треба порахувати суму з ПДВ для кожного з 500 замовлень. Копіювати той самий рядок коду 500 разів — не варіант. Саме для цього існують **цикли**: ви описуєте дію **один раз**, а Python повторює її для кожного елемента.

Це одна з найважливіших конструкцій для аналітика. Обробити список продажів, пройтися по всіх містах, перевірити кожен рядок даних — усе це цикли.

У Python їх два:
- **`for`** — коли наперед відомо, що саме перебираємо (список, рядок, словник)
- **`while`** — коли повторюємо, доки виконується певна умова

На практиці `for` використовується значно частіше, але розуміти треба обидва.

Синтаксис циклу for:
```

    тіло циклу for
```

Тіло циклу for завжди відділяється відступом. Наприклад:

````python
names = ['Monica', 'Ross', 'Chandler', 'Joey', 'Rachel']
for current_name in names:
    print(current_name)
````

Ітерація (iteration) – це одне коло повторення циклу.  
`names` у прикладі вище - об'єкт, що ітерується (iterable) - об'єкт, який віддає по одному результату за кожну ітерацію.

Тепер аби нам оновити те, що хочемо виводити на екран, нам треба лише внести зміну в один рядок тіла циклу:

````python
names = ['Monica', 'Ross', 'Chandler', 'Joey', 'Rachel']
for current_name in names:
    print(f"Hello, {current_name}! Your name length is {len(current_name)}.")
````

### Функція range
Функція range() повертає послідовність цілих чисел з урахуванням переданих їй аргументів. Синтаксис функції range():
```
range(stop)
range(start, stop[, step])
```

Аргумент start – це перше значення в діапазоні. Якщо функція range() викликається лише з одним аргументом, Python вважає, що start = 0.
Аргумент stop – це верхня межа діапазону. Важливо розуміти, що саме граничне значення не включається до послідовності.

У прикладі нижче у нас є діапазон, що починається зі значення за умовчанням, що дорівнює 0, і включає цілі числа менше 6.

````python
for i in range(6):
    print(i, end="; ")
````

````python
# передамо два аргументи, при чому перший - відʼємний
for i in range(-2, 4):
    print(i, end="; ")
````

````python
# а так працювати не буде, бо за замовчанням step=1
for i in range(-5, -10):
    print(i, end="; ")
````

````python
# натомість так буде
for i in range(-5, -10, -1):
    print(i, end="; ")
````

````python
for i in range(200, 300, 4):
    print(i, end="; ")
````

````python
number = 3
result = 1

for i in range(6):
  result = result * number
  print(f"Number {number} to power {i+1} is {result}")
````

Що робить range(6)

Генерує 6 чисел: 0, 1, 2, 3, 4, 5. Тобто цикл виконається рівно 6 разів. Сама змінна i тут навіть не використовується всередині циклу — вона потрібна лише для того, щоб задати кількість повторень.

Ключова ідея

Це накопичувальний патерн (accumulator pattern) — на кожній ітерації result перезаписується на основі свого ж попереднього значення:

````python
result = result * number
````

Читай це як "візьми поточний result, помнож на number, і збережи назад у result". Саме тому значення постійно ростуть — кожен наступний крок множить уже накопичений результат, а не початкову 1.

Що це фактично обчислює

Це — обчислення степенів числа 3, крок за кроком:

````python
# 3¹ = 3
# 3² = 9
# 3³ = 27
# 3⁴ = 81
# 3⁵ = 243
# 3⁶ = 729
````

Тобто код вручну (без **) реалізує те, що можна було б записати одним рядком як number ** 6, але тут показує всі проміжні кроки множення, а не лише фінальний результат.

Важливо розуміти

Якби result перед циклом дорівнював 0 замість 1, результат завжди був би 0 (бо 0 * будь-що = 0). Тому для множення накопичувач завжди стартує з 1 (нейтральний елемент множення), а не з 0 (це нейтральний елемент для додавання, як у прикладі з сумою чисел раніше).

## While
While - цикл з постумовою. Синтаксис:
```
while умова:
    тіло циклу
```
Розглянемо приклад:

````python
i = 1
while i < 5:
    print(i)
    i += 1
````

````python
i
````

`i = 1` - змінна, яка є лічильником для циклу while.  
Далі йде позначення циклу – while та умова виконання циклу – `i < 5`.  
Т.я. `i = 1`, то умова правильна `1 < 5` (True) і цикл починає виконуватися.  
Виводиться значення змінної `i` (`1`), далі значення змінної `i` збільшується на 1 цикл повторюється спочатку.  

Тепер `i = 2` т.я. ми збільшили її на 1 в останній інструкції циклу.  
Перевіряється умова `2 < 5`, умова правильна (True) і цикл продовжується.  
Друкується значення змінної `i` (`2`), далі значень змінної `i` знову збільшується на 1 і тепер `i = 3`.  

Знову відбувається перевірка умови циклу і так по колу.  

Після того, як умова не виконується, цикл зупиняється.  

````python
i = 1
while i < 5:
    print(f'Умова виконується i = {i}, {i} < 5')
    i += 1
````

### Управління циклом з break та continue
Оператори break та continue працюють в циклі while аналогічно до того, як в циклі for. Розглянемо візуалізацію їх роботи та приклади:

![](https://drive.google.com/uc?export=view&id=1EgY_f_rdPaSmtYBJuEMSKCHnTrWeNItp)

````python
# приклад break
i = 1
while True:     # умова завжди правдива
    if i == 5:  # якщо i рівне 5 , то викликаємо оператор break
        break   # оператор break перериває виконання циклу
    print(i)
    i += 1
````

````python
# приклад continue
i = 1
while i <= 10:
    if i % 2 != 0:
        i += 1
        continue
    print(i)
    i += 1
````

### Вкладені цикли while

Цикли можна розмішувати один в одному, це стосується як цикла while, так і for. Наприклад, виведемо табличку множення.

````python
N = 3
a = 1
while a <= N:
    b = 1
    while b <= N:
        print(a * b, end="\t")
        b += 1
    a += 1
    print("")
````

Те саме можна вивести з допомогою цикла for:

````python
max_value = 3

for a in range(1, max_value + 1):
    for b in range(1, max_value + 1):
        print(a * b, end="\t")
    print("")
````

````python
# Приклад: чому список порожній?
data = []
print(f"DEBUG: data на початку = {data}")

for i in range(-5, 5):
    if i > 10:  # Помилка в логіці!
        data.append(i)
    print(f"DEBUG: ітерація {i}, data = {data}")

print(f"DEBUG: фінальний data = {data}")  # []
````

````python
# Приклад: перевірка типів даних
def process_data(value):
    print(f"DEBUG: value = {value}, type = {type(value)}")

    if isinstance(value, str):
        print("DEBUG: це рядок, конвертуємо в число")
        value = int(value)

    result = value * 2
    print(f"DEBUG: result = {result}")
    return result

process_data("5")
process_data(10)
````

Зазвичай просто додаємо проміжні print і цього достатньо аби виявити помилки.

### Спосіб 2: Використання бібліотеки pdb (Python Debugger)

Для складніших випадків можна використовувати вбудований дебагер:


````python
import pdb

def calculate_average(numbers):
    # Встановлюємо breakpoint - виконання зупиниться тут
    pdb.set_trace()

    total = 0
    for num in numbers:
      total += num
    count = len(numbers)
    average = total / count

    return average

# При виконанні відкриється інтерактивний дебагер
# Команди:
# n (next) - наступний рядок
# c (continue) - продовжити до наступного breakpoint
# p variable - показати значення змінної
# q (quit) - вийти з дебагера

result1 = calculate_average([10, 20, 30, 40])
result2 = calculate_average([101, 201, 301, 401])
print(result1, result2)
````

### Спосіб 3: Jupyter магічна команда %debug

В Jupyter Notebook можна викликати дебагер після помилки:

````python
# Код з помилкою
def divide_numbers(a, b):
    return a / b

numbers = [10, 20, 30, 0, 40]
results = []

for num in numbers:
    result = divide_numbers(100, num)  # Помилка на num=0!
    results.append(result)

# Після помилки виконайте в новій клітинці:
# %debug
# Це відкриє дебагер в місці помилки
````
